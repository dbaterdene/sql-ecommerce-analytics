```python
import pandas as pd
```

```python
# Load raw Excel data
file_path = "../data/online_retail_II.xlsx"

df = pd.read_excel(file_path)
```

```python
# Basic dataset information
print("\n===== DATASET SHAPE =====")
print(df.shape)
```

```python
# Column names
print("\n===== COLUMNS =====")
print(df.columns.tolist())
```

```python
# First 5 rows
print("\n===== FIRST 5 ROWS =====")
print(df.head())
```

```python
# Data types
print("\n===== DATA TYPES =====")
print(df.dtypes)
```

```python
# Missing values
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())
```

```python
# Duplicate rows
print("\n===== DUPLICATES =====")
print(df.duplicated().sum())
```

```python
# Basic numerical statistics
print("\n===== NUMERICAL SUMMARY =====")
print(df.describe())
```

```python
# Cancellation rows
cancellation_rows = (
    df["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)
print("\n===== CANCELLATION ROWS =====")
print(cancellation_rows)
```

```python
# Negative quantity rows
negative_quantity_rows = (df["Quantity"] < 0).sum()
print("\n===== NEGATIVE QUANTITY ROWS =====")
print(negative_quantity_rows)
```

```python
# Non-positive price rows
non_positive_price_rows = (df["Price"] <= 0).sum()
print("\n===== NON-POSITIVE PRICE ROWS =====")
print(non_positive_price_rows)
```

```python
# Unique customers
unique_customers = df["Customer ID"].nunique()
print("\n===== UNIQUE CUSTOMERS =====")
print(unique_customers)
```

```python
# Unique products
unique_products = df["StockCode"].nunique()
print("\n===== UNIQUE PRODUCTS =====")
print(unique_products)
```

```python
# Unique countries
unique_countries = df["Country"].nunique()
print("\n===== UNIQUE COUNTRIES =====")
print(unique_countries) 
```
