```python
#import libaries
import pandas as pd
```

```python
# Load ALL sheets as a dictionary of DataFrames
sheets = pd.read_excel("../data/online_retail_II.xlsx", sheet_name=None)

# Count the number of rows per sheet
total_rows = 0
for name, df in sheets.items():
    print(f"Sheet: {name}, Rows: {len(df)}")
    total_rows += len(df)

# Total rows across all sheets
print("Total rows across all sheets:", total_rows)
```

```python
# Investigate missing CustomerID
total_missing = 0;

# Count the number of missing CustomerID all sheet
for name, df in sheets.items():
    missing_customer = df["Customer ID"].isna().sum()
    total_missing += missing_customer

# Calculate the percentage of missing CustomerID 
missing_customer_percentage = (
    total_missing * 100 / Total_rows
)

# Total missing customerID and percentage across all sheets
print("\n===== MISSING CUSTOMERID =====")
print(f"Total Rows missing CustomerID: {total_missing:,}")
print(f"Percentage of rows: {missing_customer_percentage:.2f}%")
```

```python
# Investigate duplicate rows
total_duplicated = 0;

# Count the number of duplicate all sheet
for name, df in sheets.items():
    duplicate_sheet = df.duplicated().sum()
    print(f"Sheet: {name}, Duplicate Rows: {duplicate_sheet}")
    total_duplicated += duplicate_sheet

#Total duplicated rows across all sheets
print("\n===== DUPLICATE ROWS =====")
print(f"Total Duplicate rows: {total_duplicated:,}")
```

```python
# Investigate Cancelled Invoices
total_cancellation = 0;

#Count th number of Cancellation all sheet
for name, df in sheets.items():
    cancellation_sheet = df["Invoice"].astype(str).str.startswith("C").sum()
    print(f"sheet: {name}",cancellation_sheet)
    total_cancellation += cancellation_sheet
    

print("\n===== CANCELLATIONS =====")
print(f"Cancellation rows: {total_cancellation:,}")
```

```python
# Investigate negative quantities
total_negative_quantity = 0;

#Count the number of negative quantities all sheet
for name, df in sheets.items():
    negative_quantity_sheet = (df["Quantity"] < 0).sum()
    print(f"sheet: {name}",negative_quantity_sheet)
    total_negative_quantity += negative_quantity_sheet

print("\n===== NEGATIVE QUANTITIES =====")
print(f"Negative Quantity rows: {total_negative_quantity:,}")
```

```python
#Compare cancellation and negative quantity rows
print("Both cancellation AND negative quantity rows:", (total_cancellation & total_negative_quantity).sum())
```

```python
#Find negative quatitites that aren't cancellations
negative_quantity_not_cancellation = total_negative_quantity & ~total_cancellation
print(f"Negative quantity rows that aren't cancellations invoices: {negative_quantity_not_cancellation:,}")
```

```python
#Investigate non-positive prices
total_non_positive_price = 0

#Count the number of non-positive prices all sheet
for name, df in sheets.items():
    non_positive_price_sheet = (df["Price"] <= 0).sum()
    print(f"sheet: {name}",non_positive_price_sheet)
    total_non_positive_price += non_positive_price_sheet

print("\n===== NON-POSITIVE PRICES =====")
print(f"Non-Positive Price rows: {total_non_positive_price:,}")
```

```python
#Investigate zero-price transactions
total_zero_price = 0
total_Negative_price = 0

#Total rows with zero prices across all sheets
for name, df in sheets.items():
    zero_price_sheet = (df["Price"] == 0).sum()
    print(f"sheet: {name}",zero_price_sheet)
    total_zero_price += zero_price_sheet

print("\n===== ZERO PRICES =====")
print(f"Zero Price rows: {total_zero_price:,}\n")

#Total rows with negative prices across all sheets
for name, df in sheets.items():
    negative_price_sheet = (df["Price"] < 0).sum()
    print(f"sheet: {name}",negative_price_sheet)
    total_Negative_price += negative_price_sheet

print("\n===== NEGATIVE PRICES =====")
print(f"Negative Price rows: {total_Negative_price:,}")

```

```python
#Investigate products descriptions
total_missing_description = 0

#Count the number of missing product descriptions all sheet
for name, df in sheets.items():
    missing_description_sheet = df["Description"].isna().sum()
    print(f"sheet: {name}",missing_description_sheet)
    total_missing_description += missing_description_sheet

print("\n===== MISSING PRODUCT DESCRIPTIONS =====")
print(f"Missing Product Description rows: {total_missing_description:,}")
```

```python
#Investigate StockCode
all_codes = []

#Count the number of missing StockCode all sheet
for name, df in sheets.items():
    codes = (
        df["StockCode"]
        .astype(str)
        .drop_duplicates()
        .to_list()
    )
    all_codes.extend(codes)

# Remove duplicates across sheets
all_codes = list(set(all_codes))

print("\n===== SAMPLE STOCK CODES (ALL SHEETS) =====")
print(all_codes[:30])
```

```python
#Calculate potential revenue  revenue = sum of (Quantity * Price) for all rows
total_revenue = 0
for name, df in sheets.items():
    revenue_sheet = (df["Quantity"] * df["Price"]).sum()
    print(f"sheet: {name}, Revenue: {revenue_sheet}")
    total_revenue += revenue_sheet

#Cancellation revenue
total_cancellation_revenue = 0
for name, df in sheets.items():
    cancellation_revenue_sheet = (df.loc[df["Invoice"].astype(str).str.startswith("C"), "Quantity"] * df.loc[df["Invoice"].astype(str).str.startswith("C"), "Price"]).sum()
    print(f"sheet: {name}, Cancellation Revenue: {cancellation_revenue_sheet}")
    total_cancellation_revenue += cancellation_revenue_sheet

#Positive revenue all sheets
total_positive_revenue = 0
for name, df in sheets.items():
    positive_revenue_sheet = (df.loc[df["Price"] > 0, "Quantity"] * df.loc[df["Price"] > 0, "Price"]).sum()
    print(f"sheet: {name}, Positive Revenue: {positive_revenue_sheet}")
    total_positive_revenue += positive_revenue_sheet


print("\n===== POTENTIAL REVENUE =====")
print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Total Cancellation Revenue: {total_cancellation_revenue:,.2f}")
print(f"Total Positive Revenue: {total_positive_revenue:,.2f}")
```

```python
# Investigate negative revenue rows (Quantity * Price < 0)
total_negative_revenue_rows = 0

# Count the number of negative revenue rows across all sheets
for name, df in sheets.items():
    negative_revenue_rows_sheet = ((df["Quantity"] * df["Price"]) < 0).sum()
    print(f"sheet: {name}, Negative Revenue Rows: {negative_revenue_rows_sheet}")
    total_negative_revenue_rows += negative_revenue_rows_sheet

print("\n===== NEGATIVE REVENUE =====")
print(f"Total Negative Revenue: {total_negative_revenue_rows}")    
```
