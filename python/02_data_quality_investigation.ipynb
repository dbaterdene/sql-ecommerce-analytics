```python
#import libaries
import pandas as pd
```

```python
# load the data
file_path = "../data/online_retail_II.xlsx"
    
df = pd.read_excel(file_path)

print("Data loaded successfully!")
print(f"rows: {len(df):,}")
```

```python
# Check for missing CustomerID 
missing_customer = df["Customer ID"].isna()

# Count the number of missing CustomerID 
missing_customer_count = missing_customer.sum()

# Calculate the percentage of missing CustomerID 
missing_customer_percentage = (
    missing_customer_count / len(df) * 100
)

print("\n===== MISSING CUSTOMERID =====")
print(f"Rows missing CustomerID: {missing_customer_count:,}")
print(f"Percentage of rows: {missing_customer_percentage:.2f}%")
```
