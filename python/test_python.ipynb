```python
import pandas as pd

# Load ALL sheets as a dictionary of DataFrames
sheets = pd.read_excel("../data/online_retail_II.xlsx", sheet_name=None)

# Count rows per sheet
Total_rows = 0
for name, df in sheets.items():
    print(f"Sheet: {name}, Rows: {len(df)}")
    Total_rows += len(df)

# Total rows across all sheets
print("Total rows across all sheets:", total_rows)

```

```python

# Check for missing Customer ID values in each sheet
total_missing = 0

for name, df in sheets.items():
    missing = df["Customer ID"].isna().sum()
    print(f"Sheet: {name}, Missing Customer ID rows: {missing}")
    total_missing += missing

# Total missing Customer ID across all sheets
print("\nTOTAL missing Customer ID across all sheets:", total_missing)
```
